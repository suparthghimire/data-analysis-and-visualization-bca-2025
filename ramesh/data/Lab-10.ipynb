```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import holoviews as hv
from holoviews import opts, dim
from bokeh.plotting import show
hv.extension('bokeh')

# Sample migration data
data = {
    'origin': ['Africa', 'Africa', 'Asia', 'Asia', 'Europe', 'Europe', 'North America', 'South America'] * 3,
    'destination': ['Europe', 'North America', 'Europe', 'North America', 'North America', 'Asia', 'Europe', 'North America'] * 3,
    'year': [2015]*8 + [2018]*8 + [2020]*8,
    'count': [120000, 90000, 150000, 110000, 80000, 60000, 70000, 50000,
              130000, 95000, 145000, 105000, 85000, 55000, 75000, 55000,
              125000, 85000, 140000, 100000, 90000, 50000, 80000, 60000],
    'continent': ['Africa', 'Africa', 'Asia', 'Asia', 'Europe', 'Europe', 'North America', 'South America'] * 3
}

migration = pd.DataFrame(data)
print(migration.head())
```

### 1. Sankey Diagram


```python
from pySankey.sankey import sankey

# Aggregate data by origin and destination
flow_data = migration.groupby(['origin', 'destination'])['count'].sum().reset_index()

# Create Sankey diagram
sankey(
    left=flow_data['origin'],
    right=flow_data['destination'],
    leftWeight=flow_data['count'],
    rightWeight=flow_data['count'],
    colorDict=None,
    fontsize=10
)

plt.title("Global Migration Flows (2015-2020)")
plt.show()
```

### 2. Alluvial Plot


```python
# For alluvial plot, we'll need data across multiple stages (years)
# Let's prepare data for 3 years (2015, 2018, 2020)

# Create acyclic flows between consecutive years for each origin
years = [2015, 2018, 2020]
alluvial_data = []

for origin in migration['origin'].unique():
    for i in range(len(years)-1):
        count_from = migration[(migration['origin'] == origin) & (migration['year'] == years[i])]['count'].sum()
        count_to = migration[(migration['origin'] == origin) & (migration['year'] == years[i+1])]['count'].sum()
        # Use min to avoid artificially inflating flows
        flow_count = min(count_from, count_to)
        alluvial_data.append({
            'source': f"{origin} ({years[i]})",
            'target': f"{origin} ({years[i+1]})",
            'count': flow_count
        })

alluvial_df = pd.DataFrame(alluvial_data)

# Create alluvial plot using HoloViews (now acyclic)
alluvial = hv.Sankey(alluvial_df, kdims=['source', 'target'], vdims=['count'])
alluvial.opts(
    opts.Sankey(width=800, height=500, label_position='left',
               edge_color='source', node_color='index', cmap='Category20')
)

alluvial
```

### Chord Diagram (with Bokeh)


```python
import holoviews as hv

# Prepare data for chord diagram
flow_data = migration.groupby(['origin', 'destination'])['count'].sum().reset_index()
chord_data = flow_data.rename(columns={'origin': 'source', 'destination': 'target', 'count': 'value'})

# Create Chord diagram using Holoviews
chord = hv.Chord(chord_data)
chord.opts(
    opts.Chord(
        labels='source',
        node_color='source',
        edge_color='source',
        cmap='Category20',
        width=600,
        height=600,
        title="Migration Flow Chord Diagram"
    )
)

chord
```

##  Directed Graph (Flow Style)


```python
import networkx as nx

# Create directed graph
G = nx.DiGraph()

# Add edges with weights
for _, row in flow_data.iterrows():
    G.add_edge(row['origin'], row['destination'], weight=row['count']/10000)

# Draw the graph
pos = nx.spring_layout(G)
weights = [G[u][v]['weight'] for u,v in G.edges()]

plt.figure(figsize=(10, 8))
nx.draw(G, pos, with_labels=True, width=weights, edge_color='skyblue',
        node_color='lightgreen', node_size=2000, 
        arrows=True, arrowsize=20, font_size=12)

plt.title("Directed Graph of Migration Flows")
plt.show()
```

### Stream Graph


```python
import matplotlib.pyplot as plt
import pandas as pd

# Prepare time_data from migration DataFrame
time_data = migration[['year', 'continent', 'count']]

# Aggregate counts to avoid duplicates
agg_time_data = time_data.groupby(['year', 'continent'])['count'].sum().reset_index()

# Prepare data (pivot to wide format)
wide_data = agg_time_data.pivot(index='year', columns='continent', values='count')

plt.figure(figsize=(12, 6))
plt.stackplot(wide_data.index, wide_data.T, labels=wide_data.columns,
              colors=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd'])
plt.legend(loc='upper left')
plt.title('Migration Trends by Continent (2015-2020)')
plt.xlabel('Year')
plt.ylabel('Migration Count')
plt.show()
```
