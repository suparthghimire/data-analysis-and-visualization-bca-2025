```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    'Age':  [22, 38, 26, 35, 28, 40, 19, 45, 32, 50],
    'Fare': [7.25, 71.28, 7.92, 53.10, 8.05, 31.00, 18.00, 83.48, 26.55, 50.00]
}

df = pd.DataFrame(data)

# ------------- 1. Standard Histogram of Age -------------
plt.figure(figsize=(7, 4))
sns.histplot(df['Age'], bins=6, kde=False)   # standard histogram only
plt.title("Histogram of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# ------------- KDE (Density Plot) of Age -------------
plt.figure(figsize=(7, 4))
sns.kdeplot(df['Age'], fill=True)            # smooth density curve
plt.title("KDE (Density) Plot of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Density")
plt.tight_layout()
plt.show()

# ------------- 2. Box Plot of Fare -------------
plt.figure(figsize=(6, 3))
sns.boxplot(x=df['Fare'])
plt.title("Box Plot of Ticket Fare")
plt.xlabel("Fare")
plt.tight_layout()
plt.show()

# ------------- 3. Summary Statistics for Fare (Min, Q1, Median, Q3, Max, Outliers) -------------
Q1 = df['Fare'].quantile(0.25)
Q2 = df['Fare'].median()
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]['Fare'].tolist()

print("Fare Summary Statistics (from Box Plot logic):")
print(f"Min      : {df['Fare'].min():.2f}")
print(f"Q1 (25%) : {Q1:.2f}")
print(f"Median   : {Q2:.2f}")
print(f"Q3 (75%) : {Q3:.2f}")
print(f"Max      : {df['Fare'].max():.2f}")
print(f"IQR      : {IQR:.2f}")
print(f"Lower bound for outliers: {lower_bound:.2f}")
print(f"Upper bound for outliers: {upper_bound:.2f}")
print(f"Outliers : {outliers if outliers else 'No outliers'}")
```

## 4. Brief Analysis of Both Visualizations

### a) Histogram & KDE of Age

The **standard histogram** shows how many passengers fall into different age ranges.  
With this data, most ages are between roughly **20 and 45 years**, with fewer passengers at the extremes (19 and 50).

The **KDE (density) plot** provides a smooth curve over the age range.  
It suggests that the distribution has **one main peak around the late 20s to mid-30s**, indicating that most passengers are young to middle-aged adults.

There is no strong skew in such a small sample, but we can see that **ages are spread out** between 19 and 50, with no extremely young or extremely old passengers.

### b) Box Plot of Fare

The **box plot** summarizes the ticket prices (Fare) using the five-number summary:

- **Minimum:** ≈ 7.25  
- **Q1:** ≈ 10.54  
- **Median:** ≈ 28.78  
- **Q3:** ≈ 52.33  
- **Maximum:** ≈ 83.48  

The **IQR** is quite large (~41.79), which shows that fares vary a lot between the lower and upper middle 50% of passengers.  

The **median** is closer to Q1 than Q3, suggesting a slight **right spread** (some higher fares).  

According to the **1.5×IQR rule**, no point is classified as an outlier, meaning all fares are still considered within a reasonable range for this dataset.

---

### c) Overall Interpretation

The **Age visualizations (histogram + KDE)** tell us about *who* is traveling: mostly adults in their 20s–40s.  

The **Fare box plot** tells us about *how much they are paying*: fares range from cheap to quite expensive, with no extreme outliers but a big spread in prices.  

Together, these plots give a quick picture of the **passenger profile (age)** and **ticket pricing structure (fare)** for this flight.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ------------- Dataset -------------
data = {
    'Category': ['Apples', 'Bananas', 'Oranges', 'Grapes', 'Mangoes'],
    'Count': [50, 30, 40, 20, 60]
}
df = pd.DataFrame(data)

# ------------- 1. Bar Chart -------------
plt.figure(figsize=(7, 4))
sns.barplot(x='Category', y='Count', data=df, palette='viridis')
plt.title("Bar Chart of Fruit Counts")
plt.xlabel("Fruit Category")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# ------------- 2. Pie Chart -------------
plt.figure(figsize=(6, 6))
plt.pie(df['Count'], labels=df['Category'], autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart of Fruit Distribution")
plt.tight_layout()
plt.show()
```

### Analysis of the Visualizations

#### **Bar Chart**
- The bar chart clearly shows the **absolute counts** of each fruit.
- **Mangoes** have the highest count (60), followed by **Apples (50)** and **Oranges (40)**.
- **Bananas (30)** and **Grapes (20)** have relatively fewer units in stock.
- This visualization is best for comparing exact values between categories.

#### **Pie Chart**
- The pie chart represents **relative proportions** of fruits in the total inventory.
- **Mangoes** and **Apples** together make up more than half of the stock.
- **Grapes** contribute the smallest share.
- This chart provides a good overview of **proportionate distribution**, though it’s less precise for comparing close values.

#### **Overall Insight**
- The fruit market’s inventory is **dominated by Mangoes and Apples**.
- The **bar chart** is better for precise comparison, while the **pie chart** helps visualize **overall composition**.
