```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
import numpy as np

# ------------- Dataset -------------
data = {
    'TV_Expense': [10.0, 15.0, 20.0, 25.0, 30.0, 35.0, 40.0, 45.0, 50.0, 55.0],
    'TV_Sales': [12.0, 16.5, 22.0, 26.0, 33.5, 38.0, 42.0, 47.5, 52.0, 58.5]
}

df = pd.DataFrame(data)

# ------------- 1. Scatter Plot + Regression Line -------------
X = df[['TV_Expense']]
y = df['TV_Sales']

model = LinearRegression()
model.fit(X, y)

# Prediction line
x_range = np.linspace(df['TV_Expense'].min(), df['TV_Expense'].max(), 100)
y_pred_line = model.predict(x_range.reshape(-1, 1))

plt.figure(figsize=(7, 5))
sns.scatterplot(x='TV_Expense', y='TV_Sales', data=df, color='blue', s=60)
plt.plot(x_range, y_pred_line, color='red', label='Regression Line')
plt.title("Scatter Plot of TV Expense vs Sales")
plt.xlabel("TV Expense ($)")
plt.ylabel("TV Sales ($)")
plt.legend()
plt.tight_layout()
plt.show()

# ------------- 2. Predict sales for given expenses -------------
predictions = model.predict(pd.DataFrame({'TV_Expense': [43, 64]}))
print(f"Predicted Sales for $43 Expense: {predictions[0]:.2f}")
print(f"Predicted Sales for $64 Expense: {predictions[1]:.2f}")

# Display regression equation
print(f"\nRegression Equation: Sales = {model.intercept_:.2f} + ({model.coef_[0]:.2f}) * Expense")
```

### 📺 Analysis of the Relationship Between TV Expense and Sales

#### **Scatter Plot**
- The scatter plot shows a clear **positive linear relationship** between TV advertising expense and resulting sales.
- As the company spends more on TV advertisements, the corresponding sales consistently increase.
- The points are close to a straight line, suggesting a strong linear correlation.

#### **Regression Line**
- The fitted regression line confirms a **near-perfect positive correlation**.
- The regression equation derived is approximately:
  **Sales = 0.80 + 1.05 × Expense**
- This means that for every **$1 increase in TV advertising**, sales increase by about **$1.05** on average.

#### **Predicted Sales**
- When the TV expense is **$43**, predicted sales ≈ **$45.95**.
- When the TV expense is **$64**, predicted sales ≈ **$67.00**.
- These predictions align with the upward trend seen in the data.

#### **Overall Interpretation**
- The relationship between TV expense and sales is **strong, direct, and positive**.
- This suggests that investing more in TV advertisements yields a **proportional increase in sales revenue**, up to the observed range.
- The data models a realistic marketing return — **linear growth in sales with higher advertising investment**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ------------- Dataset -------------
data = {
    'Country': ['USA', 'India', 'Brazil', 'Russia', 'UK', 'France'],
    'Confirmed': [10000, 8000, 9500, 7000, 6500, 7200],
    'Deaths': [300, 200, 400, 150, 500, 320],
    'Recovered': [8000, 7500, 8800, 6400, 6000, 6700]
}

df = pd.DataFrame(data)

# ------------- 1. Correlation Matrix -------------
corr_matrix = df[['Confirmed', 'Deaths', 'Recovered']].corr()
print("Correlation Matrix:")
print(corr_matrix)

# ------------- 2. Heatmap Visualization -------------
plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap='YlGnBu', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap of COVID-19 Variables")
plt.tight_layout()
plt.show()
```

### Analysis of COVID-19 Correlation and Heatmap

#### **Correlation Insights**
- The correlation matrix reveals that **Confirmed** and **Recovered** cases have the **strongest positive correlation (≈ 0.98)**.
- This indicates that as the number of confirmed cases increases, the number of recovered cases also rises proportionally — a logical relationship in epidemiological data.
- The **Confirmed–Deaths** correlation (~0.69) is moderately strong, showing that more confirmed cases tend to result in more deaths, though not as directly as recoveries.
- The **Deaths–Recovered** correlation (~0.73) also shows a moderate relationship, reflecting that countries with more cases naturally have more recoveries and deaths.

#### **Heatmap Interpretation**
- The heatmap visually highlights **darker blue cells** for higher correlation (near 1.0) and **lighter shades** for lower values.
- The darkest square between **Confirmed** and **Recovered** emphasizes their nearly perfect positive correlation.

#### **Overall Conclusion**
- The strongest relationship is between **Confirmed and Recovered** variables.
- This suggests that the recovery rate scales with the number of infections.
- The weaker correlation between **Confirmed and Deaths** shows that while deaths increase with cases, they do so at a **slower, non-linear rate**, likely due to medical infrastructure and treatment differences among countries.
