```python
# Map sex to gender labels for clarity
df['gender'] = df['sex'].map({0: 'Female', 1: 'Male'})

# Group by gender and calculate mean of selected metrics
scores_by_gender = df.groupby('gender')[['age', 'resting_bp', 'cholestoral']].mean()

genders = scores_by_gender.index
age = scores_by_gender['age']
resting_bp = scores_by_gender['resting_bp']
cholestoral = scores_by_gender['cholestoral']

# Plotting the stacked bar chart
plt.figure(figsize=(8, 6))
plt.bar(genders, age, label='Age', color='skyblue')
plt.bar(genders, resting_bp, bottom=age, label='Resting BP', color='lightgreen')
plt.bar(genders, cholestoral, bottom=age + resting_bp, label='Cholesterol', color='salmon')
plt.ylabel('Average Value')
plt.title('Average Medical Metrics by Gender - Stacked Bar Chart')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Map numeric sex values to string labels
df['gender'] = df['sex'].map({0: 'Female', 1: 'Male'})

# Group by gender and calculate mean
scores_by_gender = df.groupby('gender')[['age', 'resting_bp', 'cholestoral']].mean()

genders = scores_by_gender.index
age = scores_by_gender['age']
resting_bp = scores_by_gender['resting_bp']
cholestoral = scores_by_gender['cholestoral']

x = np.arange(len(genders))  # x positions for the bars
width = 0.25

plt.figure(figsize=(9, 6))
plt.bar(x - width, age, width, label='Age', color='skyblue')
plt.bar(x, resting_bp, width, label='Resting BP', color='lightgreen')
plt.bar(x + width, cholestoral, width, label='Cholesterol', color='salmon')

plt.xlabel('Gender')
plt.ylabel('Average Value')
plt.title('Grouped Bar Chart of Medical Metrics by Gender')
plt.xticks(x, genders)
plt.legend()
plt.tight_layout()
plt.show()
```

```python

explode = [0.1 if i == 0 else 0 for i in range(len(sizes))]

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', explode=explode, startangle=140,
        colors=['skyblue', 'lightgreen', 'lightcoral', 'gold'])
plt.title('Exploded Pie Chart - Chest Pain Type Distribution')
plt.axis('equal')
plt.show()
```

```python
# Inner ring: sex
inner_labels = df['sex'].value_counts().index.tolist()
inner_sizes = df['sex'].value_counts().values.tolist()
inner_colors = ['#66c2a5', '#fc8d62']

# Outer ring: chest_pain_type distribution within each sex
outer_labels = []
outer_sizes = []
outer_colors = []

cp_colors = {
    0: '#8da0cb',
    1: '#e78ac3',
    2: '#a6d854',
    3: '#ffd92f'
}

for sex in inner_labels:
    cp_counts = df[df['sex'] == sex]['chest_pain_type'].value_counts()
    for cp in [0, 1, 2, 3]:
        count = cp_counts.get(cp, 0)
        outer_labels.append(f'Sex {sex} - CP {cp}')
        outer_sizes.append(count)
        outer_colors.append(cp_colors[cp])

# Plot nested donut chart
fig, ax = plt.subplots(figsize=(8, 8))

ax.pie(outer_sizes,
       radius=1.3,
       labels=None,
       colors=outer_colors,
       wedgeprops=dict(width=0.3, edgecolor='white'))

ax.pie(inner_sizes,
       radius=1,
       labels=None,
       colors=inner_colors,
       wedgeprops=dict(width=0.3, edgecolor='white'),
       labeldistance=0.7)

legend_elements = [
    Patch(facecolor='#66c2a5', label='Male'),
    Patch(facecolor='#fc8d62', label='Female'),
    Patch(facecolor='#8da0cb', label='CP Type 0'),
    Patch(facecolor='#e78ac3', label='CP Type 1'),
    Patch(facecolor='#a6d854', label='CP Type 2'),
    Patch(facecolor='#ffd92f', label='CP Type 3')
]

plt.legend(handles=legend_elements,
           title='Categories',
           loc='lower center',
           bbox_to_anchor=(0.5, -0.15),
           ncol=2,
           frameon=False)

plt.title('Nested Pie Chart - Sex & Chest Pain Type Distribution')
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
# Prepare data
female_scores = df[df['sex'] == 0]['cholestoral']
male_scores = df[df['sex'] == 1]['cholestoral']

# Create notched box plot
plt.figure(figsize=(8, 5))
plt.boxplot(
    [female_scores, male_scores],
    labels=['Female', 'Male'],
    notch=True,
    patch_artist=True,
    boxprops=dict(facecolor='lightblue'),
    medianprops=dict(color='red'),
)

plt.title('Notched Box Plot - Cholestoral by Sex')
plt.ylabel('Cholestoral')
plt.grid(axis='y')
plt.show()
```

```python

# Select a numeric column to simulate "group comparison" — let's use 'cholestoral' grouped by 'sex'
group1 = df[df['sex'] == 0]['cholestoral'].dropna().values
group2 = df[df['sex'] == 1]['cholestoral'].dropna().values

# Define five number summary function
def five_number_summary(data):
    return {
        "Min": np.min(data),
        "Q1": np.percentile(data, 25),
        "Median (Q2)": np.median(data),
        "Q3": np.percentile(data, 75),
        "Max": np.max(data)
    }

# Plot boxplot with custom positions
plt.figure(figsize=(8, 5))
plt.boxplot(
    [group1, group2],
    positions=[1.0, 1.2],
    widths=0.2,
    notch=True,
    patch_artist=True,
    boxprops=dict(facecolor='lightblue'),
    medianprops=dict(color='red')
)

plt.xticks([1.0, 1.3], ['Female', 'Male'])
plt.title('Notched Box Plot – Cholestoral by Sex (Closer Placement)')
plt.ylabel('Cholestoral')
plt.grid(axis='y')
plt.show()

# Compute summaries
summary1 = five_number_summary(group1)
summary2 = five_number_summary(group2)

# Check equality and print summaries
np.array_equal(group1, group2), summary1, summary2
```
