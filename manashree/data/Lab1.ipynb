```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import PillowWriter

# Create the figure for plotting
fig = plt.figure()
l, = plt.plot([], [], 'b-', label="Cosine Wave")  # Changed color to blue

# Set limits for x and y axes
plt.xlim(-5, 5)
plt.ylim(-5, 5)

# Define a new function for cosine instead of sine
def cos(x):
    return np.cos(x) * 2  # Cosine wave with amplitude of 2

# Static plot data for cosine wave (not animated)
static_xlist = np.linspace(-5, 5, 100)
static_ylist = cos(static_xlist)

# Set plot labels and title
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Cosine Wave")

# Static plot (not animated)
l.set_data(static_xlist, static_ylist)
plt.show()

# Animated plot
metadata = dict(title='Cosine Wave Animation', artist='someartist')  # Animation metadata
writer = PillowWriter(fps=20, metadata=metadata)  # Saving settings for animation

animated_xlist = []  # Initialize empty list for x values in animation
animated_ylist = []  # Initialize empty list for y values in animation

# Creating the animated cosine wave and saving it as a .gif
with writer.saving(fig, 'cosinewave.gif', 100):  # Saving animation as 'cosinewave.gif'
    for xval in np.linspace(-5, 5, 100):  # Generating x values from -5 to 5
        animated_xlist.append(xval)  # Append x value
        animated_ylist.append(cos(xval))  # Append corresponding y value from cosine function
        l.set_data(animated_xlist, animated_ylist)  # Update the plot with new data points
        writer.grab_frame()  # Capture the current frame of the animation
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter

# Create the figure and axis for plotting
fig, ax = plt.subplots()

# Set the limits for the x and y axes
ax.set_xlim(1, 250)  # x-axis represents input size (n)
ax.set_ylim(0, 150)  # y-axis represents number of operations

# Set plot title and axis labels
ax.set_title("Time Complexity Growth Comparison")
ax.set_xlabel("n (input size)")  # Input size (n)
ax.set_ylabel("Operations")      # Number of operations

# Define a range of input sizes (n)
n = np.arange(1, 250)

# Define the different time complexity functions:
y_const = np.ones_like(n) * 10              # O(1) - constant time
y_log = np.log2(n) * 20                     # O(log n)
y_linear = n                                # O(n)
y_nlogn = n * np.log2(n)                    # O(n log n)
y_quad = n**2 / 1.0                         # O(n^2)

# Create empty lines for each time complexity function
line1, = ax.plot([], [], label='O(1)', color='gray')     # O(1) - gray
line2, = ax.plot([], [], label='O(log n)', color='blue')  # O(log n) - blue
line3, = ax.plot([], [], label='O(n)', color='green')     # O(n) - green
line4, = ax.plot([], [], label='O(n log n)', color='orange') # O(n log n) - orange
line5, = ax.plot([], [], label='O(n²)', color='red')      # O(n²) - red

# Add a legend to the plot
ax.legend(loc='upper left')

# Initialize the plot for animation (empty data at the start)
def init():
    line1.set_data([], [])
    line2.set_data([], [])
    line3.set_data([], [])
    line4.set_data([], [])
    line5.set_data([], [])
    return line1, line2, line3, line4, line5

# Update function for each frame in the animation
def update(i):
    # Update each line with new data for the given frame
    line1.set_data(n[:i], y_const[:i])
    line2.set_data(n[:i], y_log[:i])
    line3.set_data(n[:i], y_linear[:i])
    line4.set_data(n[:i], y_nlogn[:i])
    line5.set_data(n[:i], y_quad[:i])
    return line1, line2, line3, line4, line5

# Create the animation object using FuncAnimation
ani = FuncAnimation(fig, update, frames=len(n), init_func=init, blit=True, interval=20)

# Save the animation as a GIF file
ani.save("time_complexity.gif", writer=PillowWriter(fps=20))
```
