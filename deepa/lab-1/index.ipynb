```python
# Creating a figure
plt.figure(figsize=(4,4))
# Set title
plt.title("Line plot")
# Set label
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
# Show a plot
plt.show()
```

```python
# Multiple Lines in One Plot
# Comparing two or more related datasets is easy with multiple line plots.
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

x2 = [1, 2, 3, 4, 5]
y2 = [6, 7, 8, 9, 10]
plt.plot(x, y, label='Line 1')
plt.plot(x2, y2, label='Line 2', linestyle='--')
plt.title('Multiple Lines')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.show()
```

```python
# Scatter Plots
# Scatter plots show the relationship between two continuous variables
x_scatter = [5, 9, 8, 9, 2, 17, 2, 9, 4, 11]
y_scatter = [99, 86, 89, 88, 100, 86, 103, 89, 94, 78]
plt.scatter(x_scatter, y_scatter, color='red')
plt.title('Scatter Plot')
plt.xlabel('X Value')
plt.ylabel('Y Value')
plt.show()
```

```python
provinces = ['Province 1', 'Province 2', 'Province 3', 'Province 4', 'Province 5']
populations = [5.9, 4.8, 3.7, 2.6, 3.0]

# Unique colors for each bar
colors = ['#FF6F61', '#6B5B95', '#88B04B', '#F7CAC9', '#92A8D1']

# Create bar chart
plt.bar(provinces, populations, color=colors)

# Add title and axis labels
plt.title('Population of Provinces in Nepal')
plt.xlabel('Province')
plt.ylabel('Population (millions)')
plt.show()
```

```python
provinces = ['Province 1', 'Province 2', 'Province 3', 'Province 4', 'Province 5']
populations = [5.4, 4.2, 3.7, 2.8, 3.0]

explode = [0] * len(populations)  # start with no explosion
largest_index = populations.index(max(populations))
explode[largest_index] = 0.1  # explode the largest slice

plt.pie(populations, labels=provinces, autopct='%1.1f%%', explode=explode, startangle=140)

# Add title
plt.title('Population Distribution by Province')

# Show plot
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate a date range (30 days)
dates = pd.date_range(start='2025-05-01', periods=30)

# Step 2: Generate random temperatures (between 18°C and 35°C)
temperatures = np.random.uniform(low=59, high=95, size=30)

# Step 3: Create a DataFrame
df = pd.DataFrame({
    'Date': dates,
    'Temperature (°C)': temperatures
})

# Step 4: Plot the time series
plt.figure(figsize=(10, 5))
plt.plot(df['Date'], df['Temperature (°C)'], marker='o', color='green', linewidth=2)

# Step 5: Format the x-axis
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.title('Temperature Time Series (30 Days)')
plt.xticks(rotation=45)
plt.tight_layout()  # Adjust layout to prevent label cutoff

# Show the plot
plt.show()
```
