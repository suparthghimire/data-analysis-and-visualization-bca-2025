```python
#1.2. Basic Scatter Plot
import pandas as pd
import matplotlib.pyplot as plt

# Sample data for demonstration
data = {
    'feature1': [1, 2, 3, 4, 5, 6],
    'feature2': [2, 3, 5, 7, 11, 13]
}
df = pd.DataFrame(data)

num1 = 'feature1'
num2 = 'feature2'

plt.figure(figsize=(8,6))
plt.scatter(df[num1], df[num2], alpha=0.7)
plt.xlabel(num1)
plt.ylabel(num2)
plt.title('Basic Scatter Plot')
plt.show()

```

```python
#1.3. Bubble Chart (third variable encoded as size or color)
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Sample data
data = {
    'feature1': np.random.rand(30) * 100,
    'feature2': np.random.rand(30) * 50,
    'size_var': np.random.rand(30) * 20 + 5,   # sizes between ~5 and 25
    'color_var': np.random.rand(30) * 100      # color scale 0-100
}

df = pd.DataFrame(data)

num1 = 'feature1'
num2 = 'feature2'
size_var = 'size_var'
color_var = 'color_var'

plt.figure(figsize=(8,6))
scatter = plt.scatter(
    df[num1], df[num2],
    s=df[size_var]*10,        # scale sizes for visibility
    c=df[color_var],
    cmap='viridis', alpha=0.6, edgecolors='w', linewidth=0.5
)
plt.xlabel(num1)
plt.ylabel(num2)
plt.title('Bubble Chart')
plt.colorbar(scatter, label=color_var)
plt.show()

```

```python
#1.4. Grouped / Color-Coded Scatter Plot Using Categorical Variable
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Sample data
np.random.seed(42)
df = pd.DataFrame({
    'feature1': np.random.rand(100) * 50,
    'feature2': np.random.rand(100) * 100,
    'category': np.random.choice(['A', 'B', 'C'], 100)
})

num1 = 'feature1'
num2 = 'feature2'
cat = 'category'

plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x=num1, y=num2, hue=cat, palette='Set1', alpha=0.7)
plt.title('Grouped Scatter Plot by Category')
plt.show()
```

```python
#1.5. Regression Line - Manual Calculation (Least Squares Method)
x = df[num1].values
y = df[num2].values

# Calculate means
x_mean = np.mean(x)
y_mean = np.mean(y)

# Calculate slope (m)
m = np.sum((x - x_mean)*(y - y_mean)) / np.sum((x - x_mean)**2)

# Calculate intercept (b)
b = y_mean - m*x_mean

print(f"Regression line: y = {m:.3f}x + {b:.3f}")

# Plot
plt.figure(figsize=(8,6))
plt.scatter(x, y, alpha=0.7, label='Data points')
plt.plot(x, m*x + b, color='red', label=f'Regression line: y={m:.3f}x + {b:.3f}')
plt.xlabel(num1)
plt.ylabel(num2)
plt.title('Scatter Plot with Manual Regression Line')
plt.legend()
plt.show()
```

```python
#1.6. Regression Line Using sklearn.linear_model.LinearRegression
X = df[[num1]].values  # 2D array
y = df[num2].values

model = LinearRegression()
model.fit(X, y)

m_sklearn = model.coef_[0]
b_sklearn = model.intercept_

print(f"Sklearn Regression line: y = {m_sklearn:.3f}x + {b_sklearn:.3f}")

# Predict example
x_pred = np.array([[df[num1].min()], [df[num1].max()]])
y_pred = model.predict(x_pred)

plt.figure(figsize=(8,6))
plt.scatter(X, y, alpha=0.7, label='Data points')
plt.plot(x_pred, y_pred, color='green', label=f'Sklearn Regression line: y={m_sklearn:.3f}x + {b_sklearn:.3f}')
plt.xlabel(num1)
plt.ylabel(num2)
plt.title('Scatter Plot with Sklearn Regression Line')
plt.legend()
plt.show()

```

```python
#2. Line Chart Visualizations
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Use seaborn style
sns.set(style="whitegrid")

# Create a sample time series dataset
np.random.seed(0)
dates = pd.date_range(start="2023-01-01", periods=12, freq='M')
sales_A = np.random.randint(100, 200, size=12)
sales_B = np.random.randint(80, 160, size=12)
sales_C = np.random.randint(90, 180, size=12)

df = pd.DataFrame({
    'Date': dates,
    'Product A': sales_A,
    'Product B': sales_B,
    'Product C': sales_C
})
```

```python
#2.1. Basic Line Chart
plt.figure(figsize=(10, 5))
plt.plot(df['Date'], df['Product A'], marker='o', color='blue')
plt.title("Basic Line Chart - Product A Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.ylim(0, 250)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#2.2. Multi-Series Line Chart
plt.figure(figsize=(10, 5))
plt.plot(df['Date'], df['Product A'], label='Product A', marker='o')
plt.plot(df['Date'], df['Product B'], label='Product B', marker='s')
plt.plot(df['Date'], df['Product C'], label='Product C', marker='^')
plt.title("Multi-Series Line Chart - Sales Comparison")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.ylim(0, 250)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

```

```python
#2.3. Stacked Area Chart
plt.figure(figsize=(10, 5))
plt.stackplot(df['Date'], df['Product A'], df['Product B'], df['Product C'], 
              labels=['Product A', 'Product B', 'Product C'], alpha=0.7)
plt.title("Stacked Area Chart - Cumulative Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.ylim(0, 500)
plt.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#2.4. Spline / Smoothed Line Chart
from scipy.interpolate import make_interp_spline

# Spline interpolation for smooth line
x = np.arange(len(df))
x_smooth = np.linspace(x.min(), x.max(), 300)
spl = make_interp_spline(x, df['Product A'], k=3)
y_smooth = spl(x_smooth)

plt.figure(figsize=(10, 5))
plt.plot(x_smooth, y_smooth, color='purple', label='Product A (Smoothed)')
plt.scatter(x, df['Product A'], color='purple', marker='o')
plt.title("Spline / Smoothed Line Chart - Product A")
plt.xlabel("Time Point")
plt.ylabel("Sales")
plt.ylim(0, 250)
plt.grid(True)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#2.5. Step Line Chart
plt.figure(figsize=(10, 5))
plt.step(df['Date'], df['Product A'], where='mid', label='Product A', color='green')
plt.title("Step Line Chart - Product A Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.ylim(0, 250)
plt.grid(True)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#3.1. Basic Matrix Heatmap
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Create sample 2D data
np.random.seed(0)
data = pd.DataFrame(
    np.random.rand(6, 6),
    columns=list('ABCDEF'),
    index=list('UVWXYZ')
)

plt.figure(figsize=(8, 6))
sns.heatmap(data, annot=True, cmap='YlGnBu', linewidths=0.5)
plt.title("Basic Matrix Heatmap")
plt.show()

```

```python
#3.2. Correlation Heatmap (Pandas .corr())
# Simulated sales dataset
sales_df = pd.DataFrame({
    'A': np.random.randint(100, 200, 20),
    'B': np.random.randint(80, 160, 20),
    'C': np.random.randint(50, 150, 20),
    'D': np.random.randint(70, 130, 20)
})

plt.figure(figsize=(7, 6))
sns.heatmap(sales_df.corr(), annot=True, cmap='coolwarm', center=0)
plt.title("Correlation Heatmap")
plt.show()

```

```python
#3.3. Custom Correlation Heatmap (Manual Pearson)
# Manual Pearson correlation matrix
from scipy.stats import pearsonr

corr_manual = pd.DataFrame(index=sales_df.columns, columns=sales_df.columns)

for col1 in sales_df.columns:
    for col2 in sales_df.columns:
        corr_manual.loc[col1, col2] = pearsonr(sales_df[col1], sales_df[col2])[0]

corr_manual = corr_manual.astype(float)

plt.figure(figsize=(7, 6))
sns.heatmap(corr_manual, annot=True, cmap='RdBu', center=0)
plt.title("Manual Pearson Correlation Heatmap")
plt.show()
```

```python
#3.6. Geographic Heatmap (Non-Spatial using Plotly)
# Example: Country-level COVID-style data
import pandas as pd
import plotly.express as px

# Sample data: countries and their case counts
geo_data = pd.DataFrame({
    'Country': ['United States', 'India', 'Brazil', 'Russia', 'France', 'China'],
    'Cases': [50000000, 45000000, 30000000, 20000000, 15000000, 12000000]
})

# Create the choropleth map
fig = px.choropleth(
    geo_data,
    locations='Country',            # column with country names
    locationmode='country names',   # interpret locations as country names
    color='Cases',                  # color intensity based on cases
    hover_name='Country',           # show country name on hover
    color_continuous_scale='Reds',  # red color scale
    title='Geographic Heatmap - COVID-style Case Intensity'
)

# Save map to HTML (view in browser)
fig.write_html('geo_heatmap.html')

# Optional: auto-open in browser
import webbrowser
webbrowser.open('geo_heatmap.html')
```
