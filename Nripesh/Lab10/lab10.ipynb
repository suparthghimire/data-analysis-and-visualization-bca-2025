```python
import plotly.graph_objects as go
import pandas as pd
import plotly.express as px
import holoviews as hv
from holoviews import opts
import matplotlib.pyplot as plt
import altair as alt
from vega_datasets import data
```

## 1. Sankey Diagram
Shows flow between at least two stages using Plotly.

```python
# Sample data
labels = ["A", "B", "C", "D"]
source = [0, 1, 0, 2]
target = [2, 3, 3, 3]
value = [8, 4, 2, 8]

fig = go.Figure(data=[go.Sankey(
    node=dict(label=labels),
    link=dict(source=source, target=target, value=value)
)])
fig.update_layout(title_text="Sankey Diagram Example", font_size=10)
fig.show()
```

## 2. Alluvial Plot 


```python
df = pd.DataFrame({
    'Stage1': ['1A', '1A', '1B', '1B'],
    'Stage2': ['2A', '2B', '2A', '2B'],
    'Stage3': ['3A', '3A', '3A', '3A'],
    'Value':   [5,    3,    6,    2]
})

# Create parallel categories plot
fig = px.parallel_categories(
    df,
    dimensions=['Stage1', 'Stage2', 'Stage3'],
    color='Value',
    color_continuous_scale=px.colors.sequential.Blues
)
fig.update_layout(title="Alluvial-style Plot (Three-Stage Flow)")
fig.show()
```

## 3. Chord Diagram 

```python
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')

# Define sample connections
links = [
    ("A", "B", 5),
    ("A", "C", 3),
    ("B", "C", 2),
    ("C", "A", 4)
]

# Create Chord diagram
chord = hv.Chord(links).opts(
    opts.Chord(
        cmap='Category20',
        edge_color='target',
        node_color='index',
        edge_line_width=2,
        edge_alpha=0.7,
        labels='index',
        width=600,
        height=600,
        title="Modified Chord Diagram",
        responsive=True,
        toolbar='above'
    )
)

chord
```
