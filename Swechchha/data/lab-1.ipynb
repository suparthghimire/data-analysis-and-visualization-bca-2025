```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Line Plot – Square of values
x = np.arange(1, 11)
y = x ** 2
plt.figure()
plt.plot(x, y, marker='o')
plt.title("Square of Numbers from 1 to 10")
plt.xlabel("Number")
plt.ylabel("Square")
plt.grid(True)
plt.show()

# 2. Multiple Line Plot – sin(x) and cos(x)
x = np.linspace(0, 2 * np.pi, 100)
plt.figure()
plt.plot(x, np.sin(x), label='sin(x)', color='blue', linestyle='-')
plt.plot(x, np.cos(x), label='cos(x)', color='green', linestyle='--')
plt.title("Sine and Cosine Functions")
plt.xlabel("x (radians)")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()

# 3. Scatter Plot with Random Data
x = np.random.randn(100)
y = np.random.randn(100)
colors = ['green' if val > 0 else 'red' for val in x]
plt.figure()
plt.scatter(x, y, c=colors)
plt.title("Scatter Plot of Random Points")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.grid(True)
plt.show()
```

```python
# 4. Bar Chart – Province Populations
provinces = ['Province 1', 'Province 2', 'Bagmati', 'Gandaki', 'Lumbini']
population = [5.2, 4.5, 6.3, 2.9, 5.0]  # in millions
colors = ['blue', 'orange', 'green', 'red', 'purple']
plt.figure()
plt.bar(provinces, population, color=colors)
plt.title("Population of Provinces in Nepal")
plt.xlabel("Provinces")
plt.ylabel("Population (in millions)")
plt.xticks(rotation=45)
plt.show()

# 5. Pie Chart – Province Populations
explode = [0, 0, 0.1, 0, 0]  # Explode Bagmati
plt.figure()
plt.pie(population, labels=provinces, explode=explode, autopct='%1.1f%%', colors=colors, startangle=140)
plt.title("Population Distribution by Province")
plt.axis('equal')
plt.show()
```

```python
import pandas as pd

# 6. Temperature Time Series
dates = pd.date_range(start="2024-01-01", periods=30)
temps = np.random.normal(loc=25, scale=5, size=30)
df_temp = pd.DataFrame({'Date': dates, 'Temperature': temps})

plt.figure(figsize=(10, 5))
plt.plot(df_temp['Date'], df_temp['Temperature'], marker='o')
plt.title("Daily Temperature Over 30 Days")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import mplfinance as mpf

# 7. Candlestick Chart
# Generate dummy stock data
np.random.seed(42)
dates = pd.date_range("2024-01-01", periods=30)
open_prices = np.random.uniform(100, 110, size=30)
close_prices = open_prices + np.random.uniform(-5, 5, size=30)
high_prices = np.maximum(open_prices, close_prices) + np.random.uniform(0, 2, size=30)
low_prices = np.minimum(open_prices, close_prices) - np.random.uniform(0, 2, size=30)

df_stock = pd.DataFrame({
    'Open': open_prices,
    'High': high_prices,
    'Low': low_prices,
    'Close': close_prices
}, index=pd.DatetimeIndex(dates))

mpf.plot(df_stock, type='candle', style='charles', title='Candlestick Chart - Dummy Stock Data',
         figsize=(10, 6))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, Checkbox, Dropdown, IntSlider

def plot_functions(show_sin, show_cos, show_tan, color, num_points):
    x = np.linspace(0, 2*np.pi, num_points)
    plt.figure(figsize=(8, 4))
    
    if show_sin:
        plt.plot(x, np.sin(x), label='sin(x)', color=color)
    if show_cos:
        plt.plot(x, np.cos(x), label='cos(x)', linestyle='--', color=color)
    if show_tan:
        plt.plot(x, np.tan(x), label='tan(x)', linestyle=':', color=color)
        plt.ylim(-10, 10)
    
    plt.title("Trigonometric Functions")
    plt.xlabel("x (radians)")
    plt.ylabel("Function value")
    plt.grid(True)
    plt.legend()
    plt.show()

interact(plot_functions,
         show_sin=Checkbox(value=True, description='sin(x)'),
         show_cos=Checkbox(value=True, description='cos(x)'),
         show_tan=Checkbox(value=False, description='tan(x)'),
         color=Dropdown(options=['blue', 'green', 'red'], value='blue', description='Color'),
         num_points=IntSlider(min=50, max=200, step=50, value=100, description='Points'))
```
