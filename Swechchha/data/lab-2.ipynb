```python
import pandas as pd
from sklearn.datasets import load_iris

iris_raw = load_iris(as_frame=True)
df_raw = iris_raw.frame

print("Dataset: Iris Flower Dataset built-in from sklearn")
print("Reason: It’s simple, has numeric features, and good for visual exploration.")

# Stage 2: Parse
df = df_raw.copy()
df.columns = [col.replace(' (cm)', '').replace(' ', '_') for col in df.columns]  # Clean column names
df['species'] = df['target'].map(dict(zip(range(3), iris_raw.target_names)))  # Convert target to label
df = df.drop(columns='target')  # Drop old target

df.info()

# Stage 3: Filter
df_filtered = df[['petal_length', 'petal_width', 'species']]
print("\nFiltered data focused on petal features and species.")

# Stage 4: Mine
print("\nValue Counts:\n", df_filtered['species'].value_counts())
print("\nCorrelation:\n", df_filtered.corr(numeric_only=True))

# Stage 5: Represent
import matplotlib.pyplot as plt
import seaborn as sns

# Bar chart - average petal length per species
plt.figure(figsize=(6, 4))
sns.barplot(data=df_filtered, x='species', y='petal_length')
plt.title("Avg Petal Length per Species")
plt.show()

# Scatter plot
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df_filtered, x='petal_length', y='petal_width', hue='species')
plt.title("Petal Length vs Width by Species")
plt.show()

# Creative: Pair plot
sns.pairplot(df_filtered, hue='species')
plt.suptitle("Pair Plot of Petal Features", y=1.02)
plt.show()

# Stage 6: Refine
plt.figure(figsize=(6, 4))
sns.boxplot(data=df_filtered, x='species', y='petal_width', palette='colorblind')
plt.title("Refined: Petal Width per Species")
plt.ylabel("Petal Width (cm)")
plt.xlabel("Species")
plt.grid(True)
plt.tight_layout()
plt.show()

# Stage 7: Interact
from ipywidgets import interact

def interactive_plot(species):
    subset = df_filtered[df_filtered['species'] == species]
    plt.figure(figsize=(6, 4))
    sns.histplot(subset['petal_length'], bins=10, kde=True, color='skyblue')
    plt.title(f"Petal Length Distribution - {species}")
    plt.xlabel("Petal Length (cm)")
    plt.show()

interact(interactive_plot, species=iris_raw.target_names.tolist())
```
